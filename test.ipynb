```python
import pandas as pd
from bs4 import BeautifulSoup

df = pd.read_json('dicionario2.json')

pesquisa = "82066930019"

search_info = df.loc[df['HTML'].str.contains(pesquisa)]['HTML']

if not search_info.empty:
    for s in search_info:
        soup = BeautifulSoup(s, 'html.parser')
        print(soup.text)
else:
    print(f"Não há resultados para a {pesquisa}")
```
